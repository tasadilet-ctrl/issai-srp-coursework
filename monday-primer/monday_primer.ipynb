**The first thing that happens to any text you give an AI:** it gets chopped into pieces
called **tokens**. The model never sees letters — only these pieces. How a language gets
chopped turns out to matter a lot.

```python
# Loads in a few seconds — just the tokenizer, no model, no GPU.
from transformers import AutoTokenizer

tok = AutoTokenizer.from_pretrained("Qwen/Qwen3-0.6B")   # same model family we use in the workshop

def show(text):
    ids = tok(text, add_special_tokens=False).input_ids
    pieces = [tok.decode([i]) for i in ids]   # decode each token -> readable text
    print(f"{len(ids):>2} tokens   {text!r}")
    print("           " + "  |  ".join(repr(p) for p in pieces) + "\n")

# Watch the same meaning cost very different numbers of tokens:
show("Almaty");      show("Алматы")
show("Kazakhstan");  show("Қазақстан")
show("Hello, how are you?")
show("Сәлем, қалайсың?")
```

See it? English words are often a **fewer tokens**; the Kazakh ones **shatter into many
pieces** — and the model sees those pieces, not the word.

**Does that actually matter when you use the AI?** Maybe it copes fine — or maybe not.
